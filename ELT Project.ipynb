```python
## reading the csv
import pandas as pd
df = pd.read_csv('D:/Practise/Project ELT Dataset/netflix_titles.csv',encoding='utf-8')
```

```python
df.head(5)
```

```python
# Loading the data to SQL server
import sqlalchemy as sal
from sqlalchemy.types import NVARCHAR
engine = sal.create_engine('mssql://HEAVEN/deny?driver=ODBC+DRIVER+18+FOR+SQL+SERVER')
conn=engine.connect()
```

```python
df.to_sql('netflix_raw',con = conn,index = False,if_exists = 'replace',dtype= {'title': sal.NVARCHAR(length=200)})
conn.close()
```
